```python
import pulp
```

```python
model = pulp.LpProblem("Chi_phí_ăn_uống_thấp_nhất", pulp.LpMinimize)

x1 = pulp.LpVariable('Gạo', lowBound= 0)
x2 = pulp.LpVariable('Trứng', lowBound= 0)
x3 = pulp.LpVariable('Rau', lowBound =0)

model += 3000 * x1 + 6000 * x2 + 2000 * x3

model += 360 * x1 + 155 * x2 + 25 * x3 >= 2000
model += 7 * x1 + 13 * x2 + 2 * x3 >= 50
model += x1 >= 1
model += x2 >= 2
model += x3 >= 1

model.solve()
pulp.LpStatus[model.status]

for variable in model.variables():
  print(variable.name,"=", variable.varValue,"(g)")

print("Chi_phí_tối_ưu = ", int(pulp.value(model.objective)),"(đồng)")
```
