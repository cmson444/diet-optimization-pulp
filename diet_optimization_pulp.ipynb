```python
import pulp
```

```python
foods = {
    'Gạo':   {'cost': 3000, 'calories': 360, 'protein': 7,  'min_g': 1.0},
    'Trứng': {'cost': 6000, 'calories': 155, 'protein': 13, 'min_g': 2.0},
    'Rau':   {'cost': 2000, 'calories': 25,  'protein': 2,  'min_g': 1.0},
    'Thịt gà': {'cost': 12000, 'calories': 239, 'protein': 27, 'min_g': 1.0},
    'Sữa':     {'cost': 2500,  'calories': 60,  'protein': 3,  'min_g': 1.0}
}

requirements = {
    'calories': 2000,
    'protein': 50
}

model = pulp.LpProblem("Chi_phí_ăn_uống_thấp_nhất", pulp.LpMinimize)

food_vars = {}
for food, data in foods.items():
  food_vars[food] = pulp.LpVariable(food, lowBound=data['min_g'])

model += pulp.lpSum(data['cost'] * food_vars[food] for food, data in foods.items())

for requirement, value in requirements.items():
  model += pulp.lpSum(data[requirement] * food_vars[food] for food, data in foods.items()) >= value

model.solve()
pulp.LpStatus[model.status]

for variable in model.variables():
  print(variable.name,"=", variable.varValue,"(100 g)")

print("Chi_phí_tối_ưu =", int(pulp.value(model.objective)),"(đồng)")
```
